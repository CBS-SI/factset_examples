## 3.1 Relationships
Gets the count and percentage of overlapping products, along with the entity ID and associated company names for categories such as Supplier, Competitor, Customer, and Partner, based on the requested identifier(s).

### 3.1a `/relationships`  - Create a request object and set the parameters

```python
relationships_endpoint = 'https://api.factset.com/content/factset-supply-chain/v1/relationships'

relationships_request = {
    "data": {
        "ids": [
            "AAPL-USA"
        ],
        "relationshipType": "SUPPLIERS",
        "companyType": "PUBLIC_COMPANIES_ONLY",
        "relationshipDirection": "ALL"
    }
}

headers = {'Accept': 'application/json','Content-Type': 'application/json'}
```

### 3.1b `/relationships` - Pull data, display DataFrame properties, show initial records

#### Create a POST Request

```python
#create a post request
relationships_post = json.dumps(relationships_request)
relationships_response = requests.post(url=relationships_endpoint,
                                      data=relationships_post,
                                      auth=authorization,
                                      headers=headers,
                                      verify=False)
print('HTTP Status: {}'.format(relationships_response.status_code))
```

#### Generate a URL for a GET Request

```python
#genreate a url for a GET request
#click on URL to display data response in broswer. For request with large ids, use POST method.
get_relationships_url = relationships_endpoint + "?"
data = relationships_request["data"]

for p, v in data.items():

    if isinstance(v, list):
        get_relationships_url += p + '=' + ','.join(v) + '&'
    else:

        get_relationships_url += p + '=' + str(v) + '&'

get_relationships_url = get_relationships_url.rstrip('&')
print(get_relationships_url)
```

#### Create a Dataframe from POST

```python
#create a dataframe from POST request, show dataframe properties
relationships_data = json.loads(relationships_response.text)
relationships_df = json_normalize(relationships_data['data'])
print('COLUMNS:')
print
print(relationships_df.dtypes)
print('')
print('RECORDS:',len(relationships_df))
```
